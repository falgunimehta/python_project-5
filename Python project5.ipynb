Q1 : Write a loop to find the factorial of any number.

```python
s=int(input('your number: '))
fact=1
for i in range(1,s+1):
    fact*=i
print(fact)
```

Q2 : Print the Fibonacci series up to the 15 terms.

```python

a=0
b=1
for i in range(15):
    print(a,end=" ")
    a,b=b,a+b
```

Q3 : program to print all the prime numbers within a specified range

```python
start=int(input('your starting number: '))
end=int(input('your ending number: '))
print('prime number between your starting number', start , 'and ending number', end ,'are')
for num in range(start,end+1):
    if num>1:
        for i in range(2,num):
            if (num%i)==0:
                break
        else:
            print(num)
```

Q5 : Given a user defined number count the total number of digits in a number.

```python
num = int(input("Enter Number : "))
count = 0
while num != 0:
    num //= 10
    count+= 1
print("Total digits are: ", count)
```

Q6 : Print the numbers which are divisible by 5 and if the number is greater than 150 stop the loop iteration.

list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

```python
list1=[12,15,32,42,55,75,122,132,150,180,200]
```

```python
for i in list1:
    if i%5==0:
        print(i)
    if i>=150:
        break
```

Q7 : A student will not be allowed to appear in exam if his/her attendence is less than 75%.
Take following input from user
Number of classes held
Number of classes attended.
And print
percentage of class attended
Is student is allowed to sit in exam or not.

```python
classes=int(input('total classes: '))
attendance=int(input('classes attended:'))
attend=(attendance/classes)*100
if attend>=75:
    print('you are allowed')
else:
    print('you are not allowed')
```

Q8 : Write the frequency of character "s" in word "mississippi"

```python
a='mississippi'
c=0
for i in a:
    if 's'==i:
        c=c+1
print(c)
```
